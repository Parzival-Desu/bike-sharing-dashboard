## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Assessing Data

#### Identifying ... problem

```python
# 1. Cek missing values
print("=== Missing Values ===")
print(df.isnull().sum())

# 2. Cek baris duplikat
print("\n=== Duplicate Rows ===")
print("Jumlah baris duplikat:", df.duplicated().sum())

# 3. Cek apakah ID instant duplikat
print("\n=== Duplicate IDs ===")
print("Jumlah instant duplikat:", df["instant"].duplicated().sum())

# 4. Cek nilai kategori yang tidak sesuai
print("\n=== Validasi Kategori ===")
print("season:", sorted(df["season"].unique()))
print("yr:", sorted(df["yr"].unique()))
print("mnth:", sorted(df["mnth"].unique()))
print("hr:", sorted(df["hr"].unique()))
print("workingday:", sorted(df["workingday"].unique()))
print("weathersit:", sorted(df["weathersit"].unique()))

# 5. Cek konsistensi jumlah penyewaan
print("\n=== Konsistensi Jumlah Penyewaan ===")
print(
    "Baris dengan cnt != casual + registered:",
    (df["cnt"] != df["casual"] + df["registered"]).sum()
)

# 6. Cek rentang nilai numerik
print("\n=== Statistik Data Numerik ===")
display(df[["temp", "atemp", "hum", "windspeed", "casual",
            "registered", "cnt"]].describe())
```

```python
# Periksa apakah tanggal cocok dengan tahun, bulan, dan hari
df["dteday"] = pd.to_datetime(df["dteday"])

print("=== Pemeriksaan Konsistensi Tanggal ===")
print("Tanggal tidak cocok dengan yr:",
      (df["dteday"].dt.year != df["yr"] + 2011).sum())

print("Bulan tidak cocok dengan mnth:",
      (df["dteday"].dt.month != df["mnth"]).sum())

print("Hari dalam minggu tidak cocok:",
      (df["dteday"].dt.dayofweek != df["weekday"]).sum())

# Periksa rentang nilai yang seharusnya 0–1
print("\n=== Pemeriksaan Rentang Variabel Normalisasi ===")
for col in ["temp", "atemp", "hum", "windspeed"]:
    invalid = (~df[col].between(0, 1)).sum()
    print(f"{col}: {invalid} nilai di luar rentang 0–1")

# Periksa apakah ada penyewaan negatif
print("\n=== Pemeriksaan Nilai Negatif ===")
for col in ["casual", "registered", "cnt"]:
    print(f"{col}: {(df[col] < 0).sum()} nilai negatif")
```

```python
# Konversi hari Python:
# Senin=0 ... Minggu=6
# Dataset Bike Sharing: Minggu=0 ... Sabtu=6

weekday_bikeshare = (df["dteday"].dt.dayofweek + 1) % 7

print(
    "Tanggal dengan weekday tidak cocok:",
    (weekday_bikeshare != df["weekday"]).sum()
)
```

```python
print("=== Pemeriksaan Data ===")

print("Jumlah tanggal unik:", df["dteday"].nunique())
print("Jumlah baris:", len(df))

print(
    "Kombinasi tanggal-jam duplikat:",
    df.duplicated(subset=["dteday", "hr"]).sum()
)

print("Tanggal paling awal:", df["dteday"].min())
print("Tanggal paling akhir:", df["dteday"].max())
print("Jumlah jam unik:", df["hr"].nunique())

print("\nTipe data dteday:", df["dteday"].dtype)
print("Tipe data instant:", df["instant"].dtype)
```

```python
# Cek apakah jumlah data per tanggal sesuai 24 jam
hours_per_day = df.groupby("dteday")["hr"].nunique()

print("Tanggal dengan data kurang dari 24 jam:",
      (hours_per_day < 24).sum())

print("Tanggal dengan data lebih dari 24 jam:",
      (hours_per_day > 24).sum())

# Cek apakah setiap tanggal memiliki seluruh jam 0–23
expected_hours = set(range(24))

incomplete_dates = (
    df.groupby("dteday")["hr"]
    .apply(lambda hours: set(hours) != expected_hours)
)

print("Tanggal dengan jam yang tidak lengkap:",
      incomplete_dates.sum())
```

```python
# Distribusi jumlah jam yang tercatat per tanggal
print("=== Distribusi Jumlah Jam ===")
display(hours_per_day.value_counts().sort_index())

# Lihat beberapa tanggal yang tidak memiliki 24 jam
print("\n=== Contoh Tanggal dengan Jam Tidak Lengkap ===")
incomplete_days = hours_per_day[hours_per_day < 24]
display(incomplete_days.head(15))

# Lihat jam yang tercatat pada tiga tanggal pertama yang tidak lengkap
print("\n=== Jam yang Tercatat ===")
for date in incomplete_days.index[:3]:
    recorded = sorted(df.loc[df["dteday"] == date, "hr"].tolist())
    missing = sorted(expected_hours - set(recorded))

    print(f"Tanggal: {date.date()}")
    print(f"Jam tercatat: {recorded}")
    print(f"Jam tidak tercatat: {missing}")
    print("-" * 40)
```

### Temuan Kelengkapan Data

Pemeriksaan menunjukkan bahwa 76 dari 731 tanggal memiliki kurang dari 24 catatan per jam. Sebanyak 655 tanggal memiliki data lengkap, 62 tanggal memiliki 23 catatan, dan tanggal lainnya memiliki jumlah catatan yang lebih sedikit.

Contohnya, tanggal 2 Januari 2011 tidak memiliki catatan pada pukul 05.00, sedangkan tanggal 3 Januari 2011 tidak memiliki catatan pada pukul 02.00 dan 03.00.

**Dampak:** Analisis rata-rata penyewaan berdasarkan jam dapat memiliki jumlah observasi yang berbeda untuk setiap jam. Jam yang tidak tercatat tidak boleh dianggap sebagai jam dengan nol penyewaan tanpa bukti pendukung.

**Tindakan:** Mempertahankan data yang tersedia, mendokumentasikan ketidaklengkapan tersebut, dan memeriksa kelengkapan data sebelum membuat kesimpulan berdasarkan jam.

```python
# Hitung frekuensi setiap jam yang tidak tercatat
missing_hour_counts = {}

for date in incomplete_days.index:
    recorded = set(df.loc[df["dteday"] == date, "hr"])
    missing = expected_hours - recorded

    for hour in missing:
        missing_hour_counts[hour] = (
            missing_hour_counts.get(hour, 0) + 1
        )

missing_hour_summary = pd.Series(
    missing_hour_counts,
    name="missing_days"
).sort_values(ascending=False)

print("=== Frekuensi Jam yang Tidak Tercatat ===")
display(missing_hour_summary)
```

### Analisis Pola Catatan Jam yang Hilang

Pemeriksaan lanjutan menunjukkan bahwa jam 03.00 dan 04.00 merupakan jam yang paling sering tidak tercatat, masing-masing pada 34 tanggal. Jam 02.00 tidak tercatat pada 16 tanggal, sedangkan jam 05.00 tidak tercatat pada 14 tanggal.

**Dampak:** Ketidaklengkapan catatan dapat memengaruhi perbandingan jumlah penyewaan antarjam, khususnya pada jam-jam yang sering tidak tercatat.

**Tindakan:** Mempertahankan observasi yang tersedia tanpa mengisi catatan yang hilang dengan nol. Analisis per jam akan menggunakan data yang tersedia, dan keterbatasan kelengkapan data akan dijelaskan saat interpretasi hasil.

## Exploratory Data Analysis (EDA)

### Explore ...

```python
# Rata-rata penyewaan per jam berdasarkan hari kerja
hourly_demand = (
    df_clean.groupby(["hr", "workingday"])["cnt"]
    .mean()
    .reset_index()
)

hourly_demand["jenis_hari"] = hourly_demand["workingday"].map({
    0: "Hari nonkerja",
    1: "Hari kerja"
})

print("Rata-rata penyewaan per jam:")
display(hourly_demand.head(10))

print("\nRingkasan rata-rata penyewaan:")
display(
    hourly_demand.groupby("jenis_hari")["cnt"]
    .agg(["mean", "min", "max"])
)
```

### Analisis Pertanyaan Bisnis 1

Perbandingan rata-rata penyewaan per jam menunjukkan pola yang berbeda antara hari kerja dan hari nonkerja. Pada jam dini hari, penyewaan rata-rata cenderung lebih tinggi pada hari nonkerja. Sebaliknya, hari kerja memiliki nilai rata-rata tertinggi yang lebih besar, yang mengindikasikan adanya jam-jam tertentu dengan aktivitas penyewaan yang tinggi.

Berdasarkan hasil agregasi awal, rata-rata dari nilai penyewaan per jam adalah sekitar 192 pada hari kerja dan 181 pada hari nonkerja. Nilai puncak rata-rata per jam mencapai sekitar 525 pada hari kerja dan 373 pada hari nonkerja.

Visualisasi per jam diperlukan untuk melihat lebih jelas kapan puncak penyewaan terjadi dan bagaimana pola kedua jenis hari berbeda.

```python
weather_demand = (
    df_clean.groupby("weathersit")["cnt"]
    .agg(["mean", "median", "count"])
    .reset_index()
)

weather_labels = {
    1: "Cerah / sedikit berawan",
    2: "Berkabut / berawan",
    3: "Hujan ringan / kondisi buruk",
    4: "Cuaca buruk ekstrem"
}

weather_demand["kondisi_cuaca"] = (
    weather_demand["weathersit"].map(weather_labels)
)

print("Rata-rata penyewaan berdasarkan kondisi cuaca:")
display(weather_demand)
```

### Analisis Pertanyaan Bisnis 2

Hasil analisis menunjukkan bahwa rata-rata penyewaan sepeda berbeda berdasarkan kondisi cuaca.

* **Cerah atau sedikit berawan:** rata-rata 204,87 penyewaan, dengan 11.413 observasi.
* **Berkabut atau berawan:** rata-rata 175,17 penyewaan, dengan 4.544 observasi.
* **Hujan ringan atau kondisi buruk:** rata-rata 111,58 penyewaan, dengan 1.419 observasi.
* **Cuaca buruk ekstrem:** rata-rata 74,33 penyewaan, dengan hanya 3 observasi.

Rata-rata penyewaan tertinggi ditemukan pada kondisi cerah atau sedikit berawan, sedangkan rata-rata terendah ditemukan pada kategori cuaca buruk ekstrem. Namun, kategori terakhir hanya memiliki tiga observasi, sehingga hasilnya tidak cukup kuat untuk mewakili pola umum.

Temuan ini menunjukkan adanya hubungan antara kondisi cuaca dan jumlah penyewaan dalam dataset, tetapi tidak membuktikan bahwa cuaca merupakan satu-satunya penyebab perubahan jumlah penyewaan.

### Advanced Analysis: Binning Jam Penyewaan

```python
# Advanced Analysis: Binning Jam Penyewaan

df_binning = df_clean.copy()

df_binning["periode_hari"] = pd.cut(
    df_binning["hr"],
    bins=[-1, 5, 11, 17, 23],
    labels=[
        "Dini Hari (00–05)",
        "Pagi (06–11)",
        "Siang (12–17)",
        "Malam (18–23)"
    ]
)

period_demand = (
    df_binning.groupby("periode_hari", observed=False)["cnt"]
    .agg(["mean", "median", "count"])
    .reset_index()
)

period_demand = period_demand.rename(columns={
    "mean": "rata_rata_penyewaan",
    "median": "median_penyewaan",
    "count": "jumlah_observasi"
})

display(period_demand)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=period_demand,
    x="periode_hari",
    y="rata_rata_penyewaan"
)

plt.title("Rata-rata Penyewaan Sepeda Berdasarkan Periode Hari")
plt.xlabel("Periode Hari")
plt.ylabel("Rata-rata Jumlah Penyewaan")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

Analisis Binning Periode Hari

Jam penyewaan dikelompokkan menjadi empat periode: dini hari, pagi, siang, dan malam. Pengelompokan ini membantu membandingkan permintaan sepeda dalam rentang waktu yang lebih luas.

Hasil rata-rata penyewaan pada setiap periode dapat digunakan untuk mengidentifikasi waktu dengan permintaan relatif tinggi dan rendah. Informasi ini berpotensi membantu pengelola menentukan prioritas distribusi sepeda.

Analisis ini menggunakan data yang tersedia. Karena beberapa tanggal tidak memiliki catatan lengkap untuk seluruh 24 jam, hasilnya perlu ditafsirkan dengan mempertimbangkan keterbatasan tersebut.
